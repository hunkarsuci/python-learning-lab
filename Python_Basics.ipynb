```python
my_set = {4,5}
your_set = {4,5,6,7,8,9}
```

```python
#difference
print(my_set.difference(your_set)) # difference in my_set comparated ot your_set
```

```python
#discardprint(my_set.discard(5))
print(my_set) # discarded 5 (removed 5)
```

```python
#difference_update()
print(my_set.difference_update(your_set))
print(my_set) # differences are removed
```

```python
# intersection  also & does the intersection
print(my_set.intersection(your_set)) # only 4 and 5 remain
```

```python
# disjoint
print(my_set.isdisjoint(your_set)) # nothing in common is disjoint
```

```python
#union
print(my_set.union(your_set))
```

```python
print(my_set | your_set)
```

```python
#issubset
print(my_set.issubset(your_set))
```

```python
#issuperset
print(your_set.issuperset(my_set))
```
