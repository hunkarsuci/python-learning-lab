# Python Error Handling

Handle expected failures with exceptions and structured cleanup.

**How to use this lesson:** Run the cells in order, read the comments, and change values to test your understanding.


```python
# Error Handling 
while True:
    try:
        age = int(input('What is your age? '))
        10 / age# buggy program
    except ValueError: 
        print("!!!!")
    except ValueError: 
        print("Please enter a number")
    except ZeroDivisionError:
        print("please enter age higher than 0")
    else: # if its true state, so else stops it via break
        print('thank you')
        break
```

```python
def sum(num1, num2):
    try:
        return num1 / num2
    except (TypeError, ZeroDivisionError) as err:
        print(err)

print(sum(1, '2'))
```

```python
while True:
    try:
        age = int(input('What is your age? '))
        10 / age# buggy program
    except ValueError: 
        print("Please enter a number")
        continue # does not break the loop comes back up
    except ZeroDivisionError:
        print("please enter age higher than 0")
        break
    else: # if its true state, so else stops it via break
        print('thank you')
        break
    finally: 
        print('ok, i am finally done')
    print('can you hear me?') # we break the loop so it does not print
```
