```python
# High Order Function HOF

def greet(func): # function uses another function properties 
    func()

# HOF: map(), reduce(), filter()... 

def greet2():
    def func(): 
        return 5
    return func
```

```python
# Decorator
def my_decorator(func):
    def wrap_func(): 
        print('********')
        func()
        print('********')
    return wrap_func

@my_decorator
def hello():
    print('hello')

hello()
print()

@my_decorator
def bye():
    print('see ya later')

bye()
```

```python
# Decoretors 3 
# Decorator
def my_decorator(func):
    def wrap_func(*args,**kwargs): 
        print('********')
        func(*args,**kwargs)
        print('********')
    return wrap_func

@my_decorator
def hello(greeting, emoji = ':('):
    print(greeting, emoji)

a = my_decorator(hello)
a('hiiii')
```

```python
# Decorator in practical explanation 
from time import time
def performance(fn): # so here performance(long_time)
    def wrapper(*args, **kwargs): # we do not know the parameters of the long_time() function
        t1 = time() # time starts 
        result = fn(*args, **kwargs) # fn = long_time() and the function runs
        t2 = time() # time stops 
        print(f'took {t2-t1} s') # we compute the time difference
        return result # we return the result of the function run
    return wrapper # new decorated long_time result itself
            

@performance # we create from scratch
def long_time(): # long_time = performance(long_time) this means
    for i in range(10000000):
        i * 5

long_time()
```
