# Regular Expressions

Use the `re` module to search, match, and validate text with patterns.

**How to use this lesson:** Run the cells in order, read the comments, and change values to test your understanding.

```python
import re # regular expression module 

string = 'search this inside of this text please!'

pattern = re.compile('this') # compile the pattern 'this'

a = pattern.search(string) # search 'this' in the string
print(a)

b = pattern.findall(string) # find all occurrences of 'this' in the string
print(b) # returns a list of all occurrences of 'this' in the string

c = pattern.fullmatch(string) # check if the entire string matches the pattern 'this'
print(c) # returns None because the entire string does not match the pattern 'this'

d = pattern.match(string) # check if the beginning of the string matches the pattern 'this'
print(d) # returns None because the beginning of the string does not match the pattern 'this'

print(a.span()) # returns the start and end index of the match
print(a.start()) # returns the start index of the match
print(a.end()) # returns the end index of the match
print(a.group()) # returns the matched string
```

```python
import re 

pattern = re.compile(r"([a-zA-Z]).([a])")
string = 'search this inside of this text please! Andrei'

a = pattern.search(string) # search for the pattern in the string
b = pattern.findall(string) # find all occurrences of the pattern in the string
c = pattern.fullmatch(string) # check if the entire string matches the pattern
d = pattern.match(string) # check if the beginning of the string matches the pattern        

print(a.group(1)) # returns the first group of the match
```

```python
import re

pattern = re.compile(r"([A-Za-z0-9.]+@[\w]+\.[\w]+)")
string = 'a.boybey@keepbit.de'

a = pattern.search(string) # search for the pattern in the string
print(a)
```
