```python
# iterable 
# iterate
# generators are iterable, you can iterate over them.
# range is generator and be iterable
# list is iterable but not generator

# generator: are functions different than iterable since implementation is different 

def generator_function(num):
    for i in range(num): 
        yield i*2# we do not use return, we use yield 
                # pause the function and comes back to it

g = generator_function(10)
print(next(g)) # first time next(g) = 0 because i starts with 0
next(g) # 1 * 2 = 2
next(g) # 2 * 2 = 4

print(next(g)) # 3 * 2 = 6

# if next(g) is more than the range expiration number, so it fails (StopIteration)

#for item in generator_function(1000): # generator function loop over 1000
 #   print(item)
```

```python
def special_for(iterable): 
    iterator = iter(iterable) # iter function accepts the iterable 
    while True: 
        try: 
            print(iterator)
            print(next(iterator)) # keep call the next one 0,1,2,3)
        except StopIteration: 
            break
        
special_for([1,2,3]) # objects are in the same memory
```

```python
class MyGen():
    current = 0 # COA class object attribute
    def __init__(self, first, last): # have the option first and last
        self.first = first 
        self.last = last

    def __iter__(self): # convert from iterable to iterator
        return self # the object that I want to iterate (in class iter() = __iter__)

    def __next__(self): # next value of it 
        if MyGen.current < self.last: # we use the lop outside the class and we put inside here
            num = MyGen.current # MyGen.current starts at 0, then goes up to 100 according to definition
            MyGen.current += 1
            return num

        raise StopIteration # no more things to iterate anymore

gen = MyGen(0,100) # first is 0 and last one is 100 as first and last 
for i in gen: 
    print(i)
```

```python
# Fibonacci numbers (exercise) 
# can be done via generator or list

#generator version: 

def fib(number): # we put here the number like e.g. 20 
    a = 0 # start
    b = 1 # start
    for i in range(number): 
        yield a # give me whatever a is 
        temp = a # hold on whatever a is here
        a = b  # rule of the fibonacci that we need to know
        b = temp + b # b is equal to be temp + b 

for x in fib(1000): 
    print(x)
```

```python
# List version 
def fib2(number):
    a = 0
    b = 1
    result = []
    for i in range(number):
        result.append(a)
        temp = a
        a = b
        b = temp + b
    return result

print(fib2(1000))
```
