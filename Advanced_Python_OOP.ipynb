```python
# Polymorphism

class User(): # parent class, we use to add this class to another classes
    def sign_in(self): 
        print('logged in')

    def attack(self): # default attack method
        print('do nothing')

class Wizard(User): # inheritance: pass the parent class inside the bracket
    def __init__(self, name, power):
        self.name = name 
        self.power = power

    def attack(self):
        User.attack(self)
        print(f'attacking with power of {self.power}')

class Archer(User):
    def __init__(self, name, num_arrows):
        self.name = name 
        self.num_arrows = num_arrows

    def attack(self):
        print(f'attacking with arrows: arrows left- {self.num_arrows}')

wizard1 = Wizard('Merlin', 60) # intantiate or object
archer1 = Archer('Robin', 30) # intatiate or object
wizard1.attack() # override the attack in the original class of its. User defined there

# Polymorphism
def player_attack(char): # we create a function
    char.attack() # function calls attack method

player_attack(wizard1) # we put the object inside the function which is defined outside
player_attack(archer1)

# Polymorphism
for char in [wizard1, archer1]: 
    char.attack()
```

```python
# Dunder method (magic method)

class Toy(): 
    def __init__(self, color, age):
        self.color = color 
        self.age = age 
        self.my_dict = {'name':'Yoyo', 'has_pets':False}

    def __str__(self):
        return f'{self.color}'

    def __len__(self): 
        return 5

    def __delete__(self):
        print('deleted!!')

    def __call__(self):
        return('yess??')

    def __getitem__(self, i):
        return self.my_dict[i]

action_figure = Toy('red', 0) # instantiate or object with other name of it 
print(action_figure.__str__()) # using dunder method. action_figure.__str__()
print(str(action_figure)) # str inside the action_figure then print it 
print(len(action_figure))
print(action_figure.__len__())
print(action_figure.__delete__()) # del action_figure
print(action_figure())
print(action_figure['name'])
print(action_figure['has_pets'])
```

```python
class User(): # User is parent class 
    def sign_in(self): 
        print('logged in')

class Wizard(User): # inheritance: pass the parent class inside the bracket
    
    def __init__(self, name, power):
        self.name = name 
        self.power = power

    def attack(self):
        print(f'attacking with power of {self.power}')

class Archer(User):
    def __init__(self, name, arrows):
        self.name = name
        self.arrows = arrows

    def check_arrows(self):
        print(f'{self.arrows} remaining')

    def run(self):
        print('ran really fast')

class HybridBorg(Wizard, Archer): # wizard first and then archer inherited, so init function comes from wizard
    def __init__(self, name, power, arrows):
        Archer.__init__(self, name, arrows)
        Wizard.__init__(self, name, power)

hb1 = HybridBorg('borgie', 50, 100)
print(hb1.check_arrows())
print(hb1.run())
print(hb1.attack())
print(hb1.sign_in())
```

```python
# MRO - Method Resolution Order 

class A: 
    num = 10 
class B(A): 
    pass
class C(A):
    num = 1
class D(B, C): # D.num: first go B then C
    pass

print(D.num) # output 1 class C(A): num = 1
print(D.mro()) # .mro() gives the order-
D.__str__ # base object has str
```

```python
class A:
    num = 10

class B(A):
    pass

class C(A):
    num = 1

class D(B, C):
    pass

print(D.num)
print(D.mro())
```
