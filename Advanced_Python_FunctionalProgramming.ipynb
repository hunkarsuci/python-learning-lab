```python
# Pure Function 
def multiply_by2(li):
    new_list = []
    for item in li:
        new_list.append(item * 2)
    return new_list
    
print(multiply_by2([1,2,3]))
```

```python
#map, filter, zip and reduce 

# map
def multiply_by2(item):
    return item * 2
    
print(multiply_by2([1,2,3]))
print(map(multiply_by2, [1,2,3])) # map automatically the memory of the object gives 
print(list(map(multiply_by2, [1,2,3]))) # map take the function not with (), data
```

```python
my_list = [1,2,3]
def multiply_by2(item):
    return item * 2

print(list(map(multiply_by2, my_list))) # map takes first function then data
print(my_list) # pure function does not influence outside of the world
```

```python
# filter
my_list = [1,2,3]
def multiply_by2(item):
    return item * 2

def only_odd(item):
    return item % 2 != 0 # odd
    
print(list(filter(only_odd, my_list))) # filter also takes function , data independenly of course
```

```python
# zip 

my_list = [1,2,3]
your_list = [10,20,30]
their_list = [5,4,3]
def multiply_by2(item):
    return item * 2

def only_odd(item):
    return item % 2 != 0 # odd

# zip the two iterable like 2 lists etc. 
    
print(list(zip(my_list, your_list, their_list))) # two iterables take first item then zip together.
# 3 tuple created in a list
print(my_list) # does not change since it is defined as a list
print(your_list) # does not change since it is defined as a list
print(their_list)
```

```python
# lambda expressions 
from functools import reduce 

my_list = [1,2,3]

def only_odd(item):
    return item % 2 != 0

def accumulator(acc, item):
    print(acc, item)
    return acc + item 

print(list(filter(lambda item: item % 2 != 0, my_list))) # take the item from the mylist and multiply with 2
print(my_list)

print(reduce(lambda acc, item: acc + item, my_list))

# lambda is function can be changed in the map(), filter and reduce funtion usage for the functions 
```

```python
my_list = [5,4,2]
print(list(map(lambda item: item**2, my_list))) # list(map(lambda: lambda item: item**2, my_list))

# List Sorting 

a = [(0,2), (4,3), (9,9), (10,-1)]
print(sorted(a, key = lambda item: item[1]))
a.sort(key = lambda x: x[1])
print(a)
```

```python
# List Comprehensions

my_list = [char for char in 'hello']
my_list2 = [num for num in range(1,100)]
my_list3 = [num * 2 for num in range(1,100)]
my_list4 = [num ** 2 for num in range(1,100) if num % 2 == 1] # odd number formation with list comprehension

print(my_list)
print()
print(my_list2)
print()
print(my_list3)
print()
print(my_list4)
```

```python
# Set Comprehension
my_set = {char for char in 'hello'}
my_set2 = {num for num in range(1,100)}
my_set3 = {num * 2 for num in range(1,100)}
my_set4 = {num ** 2 for num in range(1,100) if num % 2 == 1}# odd number formation with list comprehension

print(my_set)
print()
print(my_set2)
print()
print(my_set3)
print()
print(my_set4)
```

```python
# Dictionary Comprehension
simple_dict = {'a':1, 'b':2}
my_dict = {k: v**2 for k, v in simple_dict.items() if v % 2 == 0}

my_dict2 = {num: num * 2 for num in [1,2,3]}
print(my_dict2)
```

```python
some_list = ['a','b','c','b','d','m','n','n']

duplicates = []
for value in some_list:
    if some_list.count(value) > 1:
        if value not in duplicates:
            duplicates.append(value)
            
print(duplicates)

# Comprehension way
my_list = sorted(list(set([x for x in some_list if some_list.count(x) > 1])))
print(my_list)
```
